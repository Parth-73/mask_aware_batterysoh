```python

from pathlib import Path
import pandas as pd

ROOT = Path(r"D:\battery_soh\CALCE-CX2\CX2_data")
OUTPUT = ROOT.parent / "cx2_cycle_inspection.csv"

records = []

for file in sorted(ROOT.rglob("*.xlsx")):
    try:
        book = pd.ExcelFile(file)

        for sheet in book.sheet_names:
            if not sheet.startswith("Channel"):
                continue

            df = pd.read_excel(
                book,
                sheet_name=sheet,
                usecols=[
                    "Cycle_Index",
                    "Step_Index",
                    "Current(A)",
                    "Voltage(V)",
                    "Discharge_Capacity(Ah)"
                ]
            )

            df.columns = df.columns.str.strip()
            df["Cycle_Index"] = pd.to_numeric(
                df["Cycle_Index"], errors="coerce"
            )

            df = df.dropna(subset=["Cycle_Index"])

            for cycle, group in df.groupby("Cycle_Index", sort=False):
                discharge = group[
                    group["Current(A)"] < -0.01
                ]

                if discharge.empty:
                    continue

                capacity = discharge["Discharge_Capacity(Ah)"].max()

                records.append({
                    "battery": file.parent.name,
                    "file": file.name,
                    "sheet": sheet,
                    "cycle_index": int(cycle),
                    "discharge_capacity_ah": capacity,
                    "soh_percent_candidate": capacity / 1.35 * 100,
                    "discharge_rows": len(discharge)
                })

    except Exception as e:
        print(f"Failed: {file.name}: {e}")

result = pd.DataFrame(records)
result.to_csv(OUTPUT, index=False)

print(f"Extracted {len(result)} file-level cycle records")
print(f"Saved to: {OUTPUT}")

if not result.empty:
    print(result.head(10).to_string(index=False))
```
